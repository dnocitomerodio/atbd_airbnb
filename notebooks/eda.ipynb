```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("AirbnbEDA") \
    .master("yarn") \
    .getOrCreate()

df_raw = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("quote", "\"") \
    .option("escape", "\"") \
    .option("multiLine", True) \
    .csv("hdfs://localhost:9000/data/airbnb/Listings.csv")

print(f"Filas totales: {df_raw.count()}")

df_raw.select("listing_id", "amenities", "price", "minimum_nights").show(5)
```
